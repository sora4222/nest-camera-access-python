# 4. Server login

Log in on a machine with no browser, such as a server or a remote Jupyter.

Install with the notebook tools:

```sh
pip install "googlenestcam[notebooks]"
```

Do the Google setup in [Login](../docs/login.md) first. Your OAuth client must be a **Web application** with both redirect URIs: `https://www.google.com` (Server login) and `http://localhost:8080` (browser login and SSH tunnels). Then set your Credentials as environment variables (`GOOGLENESTCAM_CLIENT_ID`, `GOOGLENESTCAM_CLIENT_SECRET`, `GOOGLENESTCAM_PROJECT_ID`) where this notebook runs.

There are three ways. Pick one.

````python
import googlenestcam as gnc
````

## Option 1: Server login

1. Run the cell below. It prints a link.
2. Open the link on any device (your laptop or phone) and approve access.
3. Google sends you to a `www.google.com` page. Copy that page's full address and paste it into the box.

Google's "enter a code on another device" login does not allow the Nest scope, so the package uses this instead ([why](../docs/adr/0001-server-login-pastes-an-address.md)).

The Token is saved to the first of: `token_path=`, `GOOGLENESTCAM_TOKEN_PATH`, `./token.json` if it exists, or `~/.config/googlenestcam/token.json`. Never commit it.

````python
gnc.login("server")
````

## Option 2: SSH tunnel

Connect to the server with the login port forwarded:

```sh
ssh -L 8080:localhost:8080 server
```

Then run `gnc.login()` on the server and open the printed link in your laptop's browser. Google's redirect to `http://localhost:8080` goes through the tunnel to the server.

## Option 3: Copy the Token

Log in on your laptop with `gnc.login()`, then either:

- copy the Token file to the server (see the paths above), or
- put the refresh token in `GOOGLENESTCAM_REFRESH_TOKEN` on the server (an `op://` 1Password reference works too).

## Check it worked

`on_missing_token="raise"` gives a `MissingTokenError` instead of trying a browser login when there is no Token, which is what you want on a server.

````python
nest = gnc.Nest(on_missing_token="raise")
for cam in nest.list_cameras():
    print(cam.name, "-", cam.kind)
````
