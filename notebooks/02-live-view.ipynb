# 2. Live view

Show a Camera's live video inside Jupyter.

Log in first with [1. Quick start](01-quick-start.ipynb).

```python
import time

from IPython.display import display

import googlenestcam as gnc

cam = gnc.list_cameras()[0]
```

## Watch for 30 seconds

`cam.stream()` keeps the newest Frame. `size=480` makes each Frame 480 pixels high, so it draws faster. Stop the cell at any time with the stop button.

```python
with cam.stream(size=480, audio=False) as stream:
    view = display(None, display_id=True)
    end = time.monotonic() + 30
    for frame in stream.frames():
        view.update(frame.to_pil())
        if time.monotonic() > end:
            break
```
