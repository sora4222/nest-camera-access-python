# 6. Several Cameras

Open a Stream on two Cameras at the same time.

Log in first with [1. Quick start](01-quick-start.ipynb).

```python
import googlenestcam as gnc

front, back = gnc.list_cameras()[:2]
print(front.name, "and", back.name)
```

## Two Streams at once

Open both in one `with` block. Each Stream keeps its own newest Frame. Here we take five Frames from each Camera, in turn.

```python
with front.stream(size=480, audio=False) as a, back.stream(size=480, audio=False) as b:
    for _, frame_a, frame_b in zip(range(5), a.frames(), b.frames(), strict=False):
        print(frame_a.time, frame_a.image.shape, "|", frame_b.time, frame_b.image.shape)
```
