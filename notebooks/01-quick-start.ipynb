````python
import googlenestcam as gnc
from PIL import Image
````

## Log in (once)

This opens your browser and saves the Token. Skip this cell on later runs. On a machine with no browser, see [4. Server login](04-server-login.ipynb).

````python
gnc.login()
````

## List your Cameras

````python
cameras = gnc.list_cameras()
for cam in cameras:
    print(cam.name, "-", cam.kind)
````

## Take a Snapshot

Pick a Camera by name, for example `gnc.camera("Front door")`. Here we take the first one. A Snapshot is a `Frame`: `frame.image` is a NumPy array (height x width x 3, RGB).

````python
cam = cameras[0]
frame = cam.snapshot()
print(frame.image.shape, frame.time)
Image.fromarray(frame.image)
````
