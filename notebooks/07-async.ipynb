# 7. Async use

Every call has an async version for code that already uses `asyncio`. Jupyter runs `await` at the top level, so no `asyncio.run` is needed here.

Log in first with [1. Quick start](01-quick-start.ipynb).

```python
import asyncio

import googlenestcam as gnc

cameras = await gnc.list_cameras_async()
cam = cameras[0]
```

## A Snapshot

```python
frame = await cam.snapshot_async(size=480)
print(frame.image.shape, frame.time)
```

## A Stream

`async with` opens it, and `async for` reads Frames.

```python
async with cam.stream_async(size=480, audio=False) as stream:
    count = 0
    async for frame in stream.frames():
        print(frame.time)
        count += 1
        if count == 5:
            break
```

## Snapshots from all Cameras at once

`asyncio.gather` runs them together, so it takes about as long as one.

```python
frames = await asyncio.gather(*(c.snapshot_async(size=480) for c in cameras))
for c, f in zip(cameras, frames, strict=True):
    print(c.name, f.image.shape)
```
